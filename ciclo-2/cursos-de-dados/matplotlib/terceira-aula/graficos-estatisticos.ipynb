```python
import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv("tips.csv")
```

```python
df.head()
```

## Histogramas

```python
px.histogram(df, x="total_bill", color="time", nbins=40,
            histnorm="probability density", marginal="box",
            title="Histograma da taxa paga no restaurante", color_discrete_sequence=["green", "orange"])
```
