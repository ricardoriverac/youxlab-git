## Rankeando um Array

```python
a = np.array([1, 2, 3])
print(a.ndim)
print(a.shape)
```

```python
a21 = a[np.newaxis, :]
print(a21.shape)
print(a21.ndim)
print(a21)
```

```python
a22 = a[:, np.newaxis]
print(a22.shape)
print(a22.ndim)
print(a22)
```

```python
print(a22[0][0])
```

## Concatenando arrays

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

c = np.concatenate((a, b))
d = np.concatenate((b, a))

print(c)
print(d)
```

## Comparando o processamento

```python
from time import process_time
lista_a = list(rng.integers(10, 100, 10000000))
lista_b = list(rng.integers(10, 100, 10000000))

c = []
c = lista_a * lista_b
```

```python
c = [ ]

t1 = process_time()
for i in range(len(lista_a)):
    c.append(lista_a[i] * lista_b[i])
t2 = process_time()
print(t2 - t1)
```

```python
a = rng.integers(10, 100, 10000000)
b = rng.integers(10, 100, 10000000)

t1a = process_time()
c = a * b
t2a = process_time()
print(t2a - t1a)
```
