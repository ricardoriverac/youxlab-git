```python
# 1) Importar a biblioteca
import pandas as pd



# 2) Ler o arquivo CSV
df_produtos = pd.read_csv("/home/youx/tudo_sobre_dados/comandos.basicos_ipynb/data/input/produtos.csv")
df_vendas = pd.read_csv("data/input/vendas_2025.csv")
df_lojas = pd.read_csv("data/input/lojas.csv")
```

```python

df_vendas['Faturamento'] = df_vendas['quantidade'] * df_produtos['preco']
df_lojas
```

```python
df_produtos
```

```python
faturamento_por_loja = df_lojas.groupby('nome_loja')['Faturamento'].sum()
faturamento_por_loja
```

```python
# Ordenar do maior para o menor faturamento
faturamento_por_loja = faturamento_por_loja.sort_values(ascending=False)


# Mostrar as 5 lojas com maior faturamento
display(faturamento_por_loja.head(5))
```

```python

df_vendas
```

```python
produtos_mais_vendidos = (
    df_vendas
    .groupby("id_produto")["quantidade"]
    .sum()
)
```

```python
produtos_mais_vendidos = produtos_mais_vendidos.sort_values(ascending=False)
produtos_mais_vendidos
```

```python
faturamento_por_mes = (
    df_vendas
    .groupby(df_vendas["data"].dt.month)["Faturamento"]
    .sum()
)
faturamento_por_mes
```
