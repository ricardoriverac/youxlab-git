8️⃣ Criação de colunas

# 8️⃣ Criação de colunas — bem explicado
📌 O que significa “criar uma coluna”?

Criar uma coluna é adicionar uma nova informação na tabela usando:

cálculos

condições

ou textos
baseados nas colunas que já existem.

📊 Antes:

preco	quantidade
10	2

📊 Depois:

preco	quantidade	faturamento
10	2	20
🧱 Estrutura básica (regra principal)
df["nova_coluna"] = alguma_coisa


🔹 df → sua tabela
🔹 "nova_coluna" → nome da coluna que você está criando
🔹 = → recebe o valor
🔹 alguma_coisa → cálculo, condição ou texto

✖️ 1. Criar coluna com cálculo
Exemplo: faturamento
df["faturamento"] = df["preco"] * df["quantidade"]


➡️ Para cada linha, o pandas:

pega o preco

multiplica pela quantidade

salva o resultado na nova coluna

🧠 Não precisa de for, o pandas faz sozinho.

➖ 2. Criar coluna com desconto
df["preco_final"] = df["preco"] - df["desconto"]


➡️ Subtrai o desconto do preço original.

✅ 3. Criar coluna com condição (True / False)
Exemplo: teve desconto?
df["teve_desconto"] = df["desconto"] > 0


Resultado:

True → teve desconto

False → não teve desconto

🔢 4. Criar coluna com if / else
Exemplo: classificar venda
df["tipo_venda"] = df["faturamento"].apply(
    lambda x: "Alta" if x > 200 else "Baixa"
)


➡️ Para cada linha:

se faturamento > 200 → "Alta"

senão → "Baixa"

📅 5. Criar coluna a partir de data
df["mes"] = df["data"].dt.month


➡️ Extrai o mês da data.

df["dia_semana"] = df["data"].dt.day_name()


➡️ Extrai o dia da semana.

🧹 6. Criar coluna tratando valores nulos
df["desconto"] = df["desconto"].fillna(0)


➡️ Substitui valores vazios por 0 antes de usar no cálculo.

🧪 Exemplo COMPLETO (igual trabalho)
