02-GRAFICO DE BARRAS


```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
df= pd.read_csv('tips.csv')
df.head()
```

```python
df.tail()
```

# GRAFICO DE BARRAS

```python
df_agrupado_pelo_dia = df.groupby('day').agg({'total_bill':np.sum}).reset_index()
```

```python
df_agrupado_pelo_dia
```

```python
px.bar(df_agrupado_pelo_dia, x= 'day', y='total_bill')
```

```python
df_agrupado_pelo_dia_e_sexo = df.groupby(['day','sex'],sort=False).agg({'total_bill':np.sum}).reset_index()
```

```python
df_agrupado_pelo_dia_e_sexo 
```

```python
px.bar(df_agrupado_pelo_dia_e_sexo ,x='day',y='total_bill',color='sex',barmode='group')
```

```python
px.bar(df_agrupado_pelo_dia_e_sexo ,x='total_bill',y='day',color='sex',barmode='group')
```

# GRAFICO DE DISPERSÃ0

```python
px.scatter(df, x='total_bill', y='tip',color='day',hover_name='time')
```

# Grafico de Pizza

```python
quantidade_de_pedidos_no_dia = df['day'].value_counts()
```

```python
quantidade_de_pedidos_no_dia
```

```python
quantidade_de_pedidos_no_dia.index
```

```python
px.pie(quantidade_de_pedidos_no_dia, names=quantidade_de_pedidos_no_dia.index,values=quantidade_de_pedidos_no_dia.values)
```

# 03. Grafico Estatísticos- Parte |

```python

import pandas as pd
import plotly.express as px
```

```python
df = pd.read_csv('tips.csv')
```

```python
df.head()
```

# 3.1.Histogramas

```python
px.histogram(
    df,
    x='total_bill',
    color='time',
    nbins=50,
    histnorm='probability density',
    marginal='box',
    title='Histograma da taxa paga no restaurante',
    color_discrete_sequence=['green', 'purple']
)
```
