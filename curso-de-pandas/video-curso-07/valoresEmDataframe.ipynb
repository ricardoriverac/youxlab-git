```python
# Aula 3
import pandas as pd
```

```python
# carregando dataset corretamente ==> neste caso usa o separador ';'.
data = pd.read_csv('./sample_data/GasPricesinBrazil_2004-2019.csv', sep=";")
```

```python
data
```

```python
data.head()
```

```python
produto_view = data['PRODUTO'] # a series retornada referente à coluna, NÃO É UMA COPIA, mas sim, uma REFERÊNCIA à coluna do dataframe
produto_view
```

```python
# Fazer uma cópia de uma series
produto_copy_bkp = data['PRODUTO'].copy() # retornando uma cópia de coluna 'PRODUTO'
produto_copy_bkp
```

```python
data['PRODUTO'] = 'Combustível' # atribuindo o valor constante 'Combustível' para linha do dataframe na coluna 'PRODUTO'
```

```python
data.head()
```

```python
produto_view
```

```python
produto_copy_bkp
```

```python
nrows, ncols = data.shape
nrows, ncols
```

```python
novos_produtos = [f'Produto {i}' for i in range(nrows)]
len(novos_produtos)
```

```python
# a quantidade de elementos da lista 'novos_produtos' é igual ao número de linhas do dataframe
data['PRODUTO'] = novos_produtos
```
