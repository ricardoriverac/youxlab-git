```python
# Criando um DataFrame

personagens_df = pd.DataFrame({
    'nome': ['Luke Skywalker', 'Yoda', 'Palpatine'],
    'idade': [16, 1000, 70],
    'peso': [70.5, 15.2, 60.1],
    'eh jedi': [True, True, False]} # o nome das colunas podem ter espaçoes
)
```

```python
personagens_df_renomeado = personagens_df.rename(columns={
    'nome': 'Nome Completo', # renomeia a coluna de nome 'nome] para 'Nome Completo'
    'idade': 'Idade'
})
```

```python
# Mostra o DataFrame com as alterações feitas acima
personagens_df_renomeado
```

```python
# Mostra o DataFrame original sem algterações
personagens_df
```

```python
# Coloca as alterações no DataFrame original
personagens_df.rename(columns={'nome': 'Nome Completo', 'idade': 'Idade'}, inplace=True)
```
