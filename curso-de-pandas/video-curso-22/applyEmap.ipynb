```python
df = pd.DataFrame({ 'A': [1, 2, 3, 4],
                    'B': [10, 20, 30, 40],
                    'C': [100, 200, 300, 400]},
                    index=['Linha 1', 'Linha 2', 'Linha 3', 'Linha 4'])
```

```python
df
```

```python
def nossa_soma(series):
  return series.sum() # retorna a soma de todos os valores de uma series
```

```python
df['SOMA(A, B, C )'] = df.apply(nossa_soma, axis=1)
df
```

```python
df.loc['Linha 5'] = df.apply(nossa_soma, axis=0)
df
```

```python
df['MEDIA(A, B, C)'] = df[['A', 'B', 'C']].apply(lambda series: series.mean(), axis=1)
df
```

```python
# aplica a lambda function abaixo para cada elemento da coluna
df['C * 2'] = df['C'].apply(lambda x: x * 2)
df
```

```python
df['A * 2'] = df['A'] * 2
df
```

```python
df = pd.DataFrame({ 'A': [1, 2, 3, 4],
                    'B': [10, 20, 30, 40],
                    'C': [100, 200, 300, 400]},
                    index=['Linha 1', 'Linha 2', 'Linha 3', 'Linha 4'])
df
```

```python
# retorna um novo dataframe com todos os elementos ao quadrado,
# poderíamos usar uma função ao invés de uma lambda fuction
df.applymap(lambda x: x ** 2)
```

```python
nomes = pd.Series(['João', 'Maria', 'José', 'Ana'])
nomes
```

```python
# retorna uma nova Series com todos os nomes com letras maiuscúlas,
# poderiamos usar uma função ao invés de uma lambda fuction
nomes.map(lambda x: x.upper())
```

```python
nomes
```

```python
# O Pandas já fornece uma série de métodos para manipulação de string,
# Assim, poderiamos usar o códgio abaixo para obter o mesmo resultado.
nomes.str.upper()
```
