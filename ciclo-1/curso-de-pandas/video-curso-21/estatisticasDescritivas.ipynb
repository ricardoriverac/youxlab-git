```python
# Aula 3
import pandas as pd
```

```python
# carregando dataset corretamente ==> neste caso usa o separador ';'.
# It seems the separator is not ';', let's try to read the file without specifying a separator first to inspect the header.
data = pd.read_csv('./sample_data/GasPricesinBrazil_2004-2019.csv', sep=';')
```

```python
data
```

```python
data.info()
```

```python
data_pre = data.copy()
```

```python
# Converte date columns to datetime objects
data_pre['DATA INICIAL'] = pd.to_datetime(data_pre['DATA INICIAL'], errors='coerce')
data_pre['DATA FINAL'] = pd.to_datetime(data_pre['DATA FINAL'], errors='coerce')
```

```python
# convertendo atributos/colunas para 'numeric'
for atributo in ['MARGEM MÉDIA REVENDA', 'PREÇO MÉDIO DISTRIBUIÇÃO', 'DESVIO PADRÃO DISTRIBUIÇÃO', 'PREÇO MÍNIMO DISTRIBUIÇÃO', 'PREÇO MÁXIMO DISTRIBUIÇÃO', 'COEF DE VARIAÇÃO DISTRIBUIÇÃO']:
  # converte a coluna (de valores string) para um tipo númerico
  # Em caso de erro na conversão (p. ex., uma string que não representa um número), um valor vazio (null / nan) será
  # atribuído no lugar
  data_pre[atributo] = pd.to_numeric(data_pre[atributo], errors='coerce')
```

```python
data_pre.info()
```

Limpeza #2

```python
mask = data_pre['PREÇO MÉDIO DISTRIBUIÇÃO'].isnull()
```

```python
data_pre[mask]
```

```python
# Nos dados originais, quais eram os valores do PREÇO MÉDIO DISTRIBUIÇÃO dos registros que agora possuem valores NaN
data[mask]
```

```python
data.info()
```

```python
# Retorna uma cópia do data frame 'data_pre' com todos os valores NaN de todas as colunas agora preenchidos com 0.
# Para alterar o próprio data frame, use o argumento 'inplace=True'.
data_pre_fill = data_pre.fillna(0)
data_pre_fill
```

```python
data_pre_fill[mask] # valores que eram NaN, agora são 0
```

```python
# retorna uma cópia do data frame 'data_pre' com todos os valores NaN das colunas:
# 'PREÇO MÉDIO DISTRIBUIÇÃO', 'DESVIO PADRÃO DISTRIBUIÇÃO', 'PREÇO MÍNIMO DISTRIBUIÇÃO' E 'MÉDIO MÁXIMO DISTRIBUIÇÃO', respectivamente,
data_pre_fill = data_pre.fillna(value={
    'PREÇO MÉDIO DISTRIBUIÇÃO': 10,
    'DESVIO PADRÃO DISTRIBUIÇÃO': 20,
    'PREÇO MÍNIMO DISTRIBUIÇÃO': 30,
    'PREÇO MÁXIMO DISTRIBUIÇÃO': 'vazio'
})
```

```python
data_pre_fill[mask]
```

```python
# remove, no próprio dataframe, todas as linhas/registros com valores NaN (vazios) em quaisquer colunas colunas/atributos.
data_pre.dropna(inplace=True)
```
