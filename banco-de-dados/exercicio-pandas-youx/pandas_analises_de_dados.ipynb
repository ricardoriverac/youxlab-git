##
 4️⃣ Criação de métricas


 4. Crie a coluna `preco_final`.
15. Crie a coluna `faturamento`.
16. Crie a coluna `teve_desconto`.
17. Crie a coluna `ticket_medio_item`.


```python
df_analise["preco_final"] = df_analise["preco_unit"] - df_analise ["desconto"]

df_analise["faturamento"] = df_analise["preco_final"] * df_analise ["qtd"]

df_analise["teve_desconto"] = df_analise["desconto"] > 0

df_analise["ticket_medio_item"] = df_analise["faturamento"] / df_analise["qtd"]
```

---
## 
 3️⃣ Limpeza e tratamento de dados

10. pd.to_
11. Padronize a coluna `loja` (remova espaços e padronize texto).
12. Substitua valores nulos de `desconto` por 0.
13. Trate valores nulos da coluna `produto`.

```python
df_analise["data"] = pd.to_datetime(df_analise["data"],format="%Y-%m-%d")




```

```python
df_analise["loja"] = (
    df_analise["loja"]
    .astype(str)
    .str.strip()
    .str.title()
)
```

```python
df_analise["desconto"] = df_analise["desconto"].fillna(0)
```

```python
df_analise["produto"] = df_analise["produto"].fillna("Não informado")
```
