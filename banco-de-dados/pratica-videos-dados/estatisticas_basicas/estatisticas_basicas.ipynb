Exemplo 2:

```python
endividamento = [25, 51, 76, 41, 55, 13]
def calcular_media(vetor):
    return sum(vetor)/len(vetor)
print(f'A média do endividamento das empresas foi: {calcular_media(endividamento)} %')
```

```python
vendas = [10, 17, 13, 15]
calcular_media(vendas)
```
